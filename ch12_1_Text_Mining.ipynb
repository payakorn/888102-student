## 12.1.5 · Text preprocessing

*step 2: lowercase, remove symbols*

The thirty reviews are tidy sentences. Real reviews are not: they shout in capitals, repeat exclamation marks and add faces made of symbols. Step 2 cleans that away before any word is counted.

> **Definition.** **Text preprocessing** prepares a text for analysis. Its two basic jobs are to **lowercase** every letter and to **remove punctuation and symbols**, so that only the words are left.

> **Why lowercase first.** *Cold*, *cold* and *COLD* are one word to a reader and three to a computer. Lowercasing first makes them one word, so every later count sees one word, not three.



#### Worked example: a messy review

The thirty reviews are too clean to show this step, so one messy review is used here instead:

*WOW!!! The noodles were GREAT :) but the delivery... 2 HOURS late?!*

- **Lowercase:** *wow!!! the noodles were great :) but the delivery... 2 hours late?!*
- **Replace every character that is not a letter, a digit or a space with a space:** the *!!!*, the *:)*, the *...* and the *?!* all go.
- **Cut at the spaces:** 11 words, *wow the noodles were great but the delivery 2 hours late*.



#### The steps

The code of this section.

*The code of this section.*



#### Lowercase

`.lower()` returns a copy of a string with every capital letter made small. Nothing else changes, so the symbols are still there.



> **Interpretation.** *WOW*, *GREAT* and *HOURS* are now *wow*, *great* and *hours*. The *!!!*, *:)*, *...* and *?!* are untouched.



```python
messy = "WOW!!! The noodles were GREAT :) but the delivery... 2 HOURS late?!"
text = messy.lower()
text
```

#### Remove the symbols and cut into words

`re.sub(r"[^a-z0-9\s]", " ", text)` replaces every character that is not a letter (a–z), a digit (0–9) or a space (\s) with a space. `.split()` then cuts the result at the spaces, so each word becomes one item of a list.



> **Interpretation.** The 11 words of the worked example. Every symbol has become a space, and the extra spaces disappear when the text is cut.



```python
cleaned = re.sub(r"[^a-z0-9\s]", " ", text)
print(cleaned)
cleaned.split()
```

#### The pattern `[^a-z0-9\s]`

The first argument of `re.sub` is a pattern: a short description of the characters to find. Read piece by piece, this one finds one character that is **not** a letter, a digit or a space:

| Piece | Meaning |
|---|---|
| `[ ... ]` | any **one** character from the list inside the brackets |
| `^` (first inside `[`) | **not**: turns the list around, so it matches every character *outside* it |
| `a-z` | any lowercase letter from a to z |
| `0-9` | any digit from 0 to 9 |
| `\s` | any space, tab or new line |

- **Without the `^`**, `[a-z0-9\s]` would find the letters, digits and spaces themselves, and replacing them would leave only the symbols.
- **The list has `a-z` but not `A-Z`**, so a capital letter counts as "not in the list" and would be removed. That is why the text is lowercased first.
- **Outside the brackets `^` means something else**: `^The` finds *The* only at the start of the text.



#### The same step with the tool

A `CountVectorizer` carries an **analyzer**: the function that lowercases a text, removes the punctuation and symbols and cuts it into words. `build_analyzer()` hands that function over so it can be used on one sentence at a time. It is stored as tokenize and used from here on.



> **Interpretation.** Ten words, the same as the hand version except one: the analyzer also drops *2*. Its rule keeps only words of two or more characters, so a single digit or letter is removed. Everything else, lowercasing and removing symbols, is what the two cells above did by hand.



```python
tokenize = CountVectorizer().build_analyzer()        # lowercase, remove symbols, split into words
tokenize(messy)
```

#### Where Section 12.1.5 leaves us

A text can now be cleaned into a list of lowercase words, by hand or with one call. Many of those words, such as *the* and *were*, say nothing about the shop. Section 12.1.6 removes them.



## 12.1.6 · Tokens and stop words

*step 3*

Section 12.1.5 cleaned a text into words. Step 3 names those words tokens and removes the ones that carry no meaning of their own.

> **Definition.** A **token** is one unit of text, usually one word. **Tokenization** cuts a text into tokens. **Filtering** removes the tokens that carry no opinion, above all the **stop words** such as *the*, *and* and *was*, which appear in almost every sentence.



#### Worked example: review 16

*The delivery took two hours and the noodles were cold.*

- **Lowercase and remove the period:** *the delivery took two hours and the noodles were cold*.
- **Cut at the spaces:** 10 tokens.
- **Remove the stop words** *the* (twice), *and*, *were* and *two*: 5 tokens remain, *delivery*, *took*, *hours*, *noodles*, *cold*.

Review 16 cut into tokens, then filtered: the four stop words are crossed out and the five words that remain say what the review is about.

*Review 16 cut into tokens, then filtered: the four stop words are crossed out and the five words that remain say what the review is about.*



#### The steps

#### The tokens of review 16

The analyzer of Section 12.1.5 is used on review 16, one of the thirty.



> **Interpretation.** Ten tokens, as in the worked example: all lowercase, the final period gone, one word per token.



```python
review_16 = reviews.loc[16, "review"]
print(review_16)
tokenize(review_16)
```

#### Without the stop words

The same analyzer built with `stop_words="english"` also removes the words on scikit-learn's English stop-word list, about three hundred of them.



> **Interpretation.** Five tokens remain: *delivery*, *took*, *hours*, *noodles* and *cold*, the five of the worked example. The opinion is still there in *cold*, but nothing has yet said whether *cold* is good or bad. That is step 4, in Part 12.2.



```python
tokenize_filtered = CountVectorizer(stop_words="english").build_analyzer()
tokenize_filtered(review_16)
```

## 12.1.7 · The bag of words

*text becomes a table of counts*

Section 12.1.6 cut one review into words. The bag of words does it for every review at once and counts the result, which is the moment text becomes a table.

> **Definition.** The **bag of words** turns text into numbers by counting: one row per text, one column per distinct word in the collection, and in each cell the number of times that word appears in that text.

Three rows of the bag of words, five of its 89 columns. A filled cell means the review used that word; the number is how many times.

*Three rows of the bag of words, five of its 89 columns. A filled cell means the review used that word; the number is how many times.*



#### What is kept and what is lost

- **Kept:** which words each review used, and how often.
- **Lost:** the order of the words. It is a bag, not a sentence: the table cannot see that review 30's *delicious* comes before a *but*. Part 12.2 puts some of that order back.
- **Gained:** every review is now a row of numbers. Summed down a column, the table says which words customers use most; handed to a classifier, the rows are features like the columns of Chapter 10.



#### The steps

The code of this section.

*The code of this section.*



#### The bag of words of the thirty reviews

`fit_transform` runs the analyzer over all thirty reviews, collects every distinct word as a column and counts. The result is stored compactly, so it is turned into a DataFrame with the words as column names.



> **Interpretation.** 30 rows by 89 columns: thirty reviews use 89 distinct words once the stop words are gone. The three rows shown are the three rows of the figure in Section 12.1.7, cell for cell.



```python
vectorizer = CountVectorizer(stop_words="english")
bag = vectorizer.fit_transform(reviews["review"])
counts = pd.DataFrame(bag.toarray(), columns=vectorizer.get_feature_names_out(), index=reviews.index)
print("bag of words:", counts.shape[0], "reviews x", counts.shape[1], "words")
counts.loc[[1, 16, 30], ["noodles", "delicious", "delivery", "cold", "late"]]
```

#### What the stop words cost

The same count without the stop-word list shows how much of the text the filtering removed: the number of distinct words, and the number of tokens in all thirty reviews together.



> **Interpretation.** With the stop words kept there are 117 distinct words and 249 tokens; without them, 89 and 142. Filtering removed 28 words from the vocabulary but 107 tokens from the text, because the few stop words appear in almost every review.



```python
bag_all = CountVectorizer().fit_transform(reviews["review"])
print("distinct words:  ", bag_all.shape[1], "with stop words,", bag.shape[1], "without")
print("tokens in total: ", bag_all.sum(), "with stop words,", bag.sum(), "without")
```

## 12.1.8 · The most frequent words

*what the customers write about*

The bag of words of Section 12.1.7 has one column per word. Summed down a column, it gives the number of times a word appears in all thirty reviews, and the words at the top of that count are what the customers write about most.

The code of this section.

*The code of this section.*



#### The twelve most frequent words

`counts.sum()` adds up each column; `sort_values` puts the most frequent first — words with the same count stay in alphabetical order — and the top twelve are drawn.



```python
word_totals = counts.sum().sort_values(ascending=False, kind="stable")   # ties stay in alphabetical order
top = word_totals.head(12)

plt.figure(figsize=(7.5, 3.6))
plt.barh(top.index[::-1], top.values[::-1], color="#2563EB")
plt.xlabel("Times the word appears in the thirty reviews")
plt.title("The twelve most frequent words")
plt.tight_layout()
plt.show()
```

#### Reading the chart

> **Interpretation.** The top of the list is what the customers write about:
>
> - **Delivery and price** lead with 6 each, then **order** with 5 and **app, food, late, noodles and service** with 4. These are the parts of the shop the customers write about; Part 12.3 scores them one by one.
> - **The times the word appears in the thirty reviews** say nothing about good or bad. *Price* could be praised or blamed; the count cannot tell.



#### The counts by star rating

The stars kept aside since Section 12.1.4 now split each count in two: the times the word appears in reviews rated 1 or 2 stars, and in reviews rated 4 or 5. `low` is a column of True/False marking the low-rated reviews.



> **Interpretation.** The same twelve words, split:
>
> - **Late: 4 and 0.** Every one of its four appearances is in a review rated 1 or 2 stars.
> - **Delivery: 4 and 2**, while **price: 2 and 4** and **food: 1 and 3** lean the other way.
> - **The count hints at an opinion without reading one.** Only the stars say which side a word is on; Part 12.2 reads the opinion from the words themselves.



```python
low = reviews["stars"] <= 2
by_stars = pd.DataFrame({"1 or 2 stars": counts[low].sum(),
                         "4 or 5 stars": counts[~low].sum()})
by_stars.loc[top.index]
```

- **The counts say what the customers talk about, not what they feel.**
